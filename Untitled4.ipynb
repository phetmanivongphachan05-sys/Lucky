## Thực hiện hồi quy đa thức và tính RMSE

### Subtask:
Vòng lặp từ bậc đa thức 1 đến 10. Đối với mỗi bậc, áp dụng PolynomialFeatures để biến đổi X_train và X_test, sau đó huấn luyện mô hình LinearRegression. Dự đoán trên X_test và tính toán RMSE giữa giá trị thực tế và giá trị dự đoán. Lưu trữ tất cả các giá trị RMSE.


**Reasoning**:
The subtask requires iterating through polynomial degrees, training a Linear Regression model for each degree, and calculating the RMSE on the test set. I will implement a loop to handle degrees from 1 to 10, using `PolynomialFeatures` to transform the data and `LinearRegression` to fit the model, then compute and store the RMSE for each iteration.



````python
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import mean_squared_error
import numpy as np

# 1. Initialize an empty list to store RMSE values
rmse_values = []

# 2. Loop through polynomial degrees from 1 to 10
for degree in range(1, 11):
    print(f"\n--- Processing Polynomial Degree: {degree} ---")

    # a. Initialize a PolynomialFeatures object with the current degree
    poly_features = PolynomialFeatures(degree=degree)

    # b. Transform X_train and X_test
    X_train_poly = poly_features.fit_transform(X_train)
    X_test_poly = poly_features.transform(X_test)

    # c. Initialize a LinearRegression object
    model = LinearRegression()

    # d. Train the LinearRegression model on X_train_poly and y_train
    model.fit(X_train_poly, y_train)

    # e. Make predictions on X_test_poly
    y_pred = model.predict(X_test_poly)

    # f. Calculate the Root Mean Squared Error (RMSE)
    rmse = np.sqrt(mean_squared_error(y_test, y_pred))
    print(f"  RMSE (Degree {degree}): {rmse:.2f}")

    # g. Add the calculated RMSE value to the rmse_values list
    rmse_values.append(rmse)

print("\nAll RMSE values:", rmse_values)
````

RMSE được tính bằng cách lấy căn bậc hai của trung bình các bình phương sai số, điều này giúp các sai số lớn được 'trừng phạt' nhiều hơn và đưa đơn vị của sai số về cùng đơn vị với biến phụ thuộc (Doanh thu).

3.  **Mục đích của việc so sánh các mô hình hồi quy đa thức với các bậc khác nhau:**
    Mục đích chính của việc so sánh các mô hình hồi quy đa thức với các bậc khác nhau là tìm kiếm sự cân bằng tối ưu giữa độ phức tạp của mô hình và khả năng khái quát hóa (hiệu suất trên dữ liệu mới, chưa thấy) của nó. Đây còn được gọi là sự đánh đổi giữa độ chệch (bias) và phương sai (variance):

4.  **Giải thích khái niệm overfitting (quá khớp) và underfitting (chưa khớp):**
    -   **Underfitting (Chưa khớp):** Xảy ra khi một mô hình quá đơn giản để nắm bắt được mối quan hệ cơ bản trong dữ liệu. Mô hình overfitting sẽ có hiệu suất rất tốt trên tập huấn luyện (RMSE thấp trên tập huấn luyện) nhưng lại kém trên tập kiểm tra (RMSE cao trên tập kiểm tra). ## Thực hiện hồi quy đa thức và tính RMSE cho cả tập huấn luyện và kiểm thử

### Subtask:
Vòng lặp từ bậc đa thức 1 đến 10. Đối với mỗi bậc, áp dụng PolynomialFeatures để biến đổi X_train và X_test, sau đó huấn luyện mô hình LinearRegression. Dự đoán trên X_train và X_test, sau đó tính toán RMSE cho cả tập huấn luyện (RMSE_train) và tập kiểm tra (RMSE_test). Lưu trữ tất cả các giá trị RMSE_train và RMSE_test.


**Reasoning**:
The subtask requires iterating through polynomial degrees from 1 to 10, training a Linear Regression model for each degree, and calculating the RMSE on both the training and test sets. I will implement a loop using `PolynomialFeatures` for data transformation and `LinearRegression` for model training, then compute and store RMSE values for both sets.



````python
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import mean_squared_error
import numpy as np

# 1. Initialize empty lists to store RMSE values for training and testing sets
rmse_train_values = []
rmse_test_values = []

# 2. Loop through polynomial degrees from 1 to 10
for degree in range(1, 11):
    print(f"\n--- Processing Polynomial Degree: {degree} ---")

    # a. Initialize a PolynomialFeatures object with the current degree
    poly_features = PolynomialFeatures(degree=degree)

    # b. Transform X_train and X_test
    X_train_poly = poly_features.fit_transform(X_train)
    X_test_poly = poly_features.transform(X_test)

    # c. Initialize a LinearRegression object
    model = LinearRegression()

    # d. Train the LinearRegression model on X_train_poly and y_train
    model.fit(X_train_poly, y_train)

    # e. Make predictions on X_train_poly and X_test_poly
    y_train_pred = model.predict(X_train_poly)
    y_test_pred = model.predict(X_test_poly)

    # f. Calculate the Root Mean Squared Error (RMSE) for training set
    rmse_train = np.sqrt(mean_squared_error(y_train, y_train_pred))
    print(f"  RMSE (Train, Degree {degree}): {rmse_train:.2f}")

    # g. Calculate the Root Mean Squared Error (RMSE) for test set
    rmse_test = np.sqrt(mean_squared_error(y_test, y_test_pred))
    print(f"  RMSE (Test, Degree {degree}): {rmse_test:.2f}")

    # h. Add the calculated RMSE values to their respective lists
    rmse_train_values.append(rmse_train)
    rmse_test_values.append(rmse_test)

print("\nAll RMSE training values:", rmse_train_values)
print("All RMSE testing values:", rmse_test_values)
````

## Thực hiện hồi quy đa thức và tính RMSE cho cả tập huấn luyện và kiểm thử

### Subtask:
Vòng lặp từ bậc đa thức 1 đến 10. Đối với mỗi bậc, áp dụng PolynomialFeatures để biến đổi X_train và X_test, sau đó huấn luyện mô hình LinearRegression. Dự đoán trên X_train và X_test, sau đó tính toán RMSE cho cả tập huấn luyện (RMSE_train) và tập kiểm tra (RMSE_test). Lưu trữ tất cả các giá trị RMSE_train và RMSE_test.


**Reasoning**:
The subtask requires iterating through polynomial degrees from 1 to 10, training a Linear Regression model for each degree, and calculating the RMSE on both the training and test sets. I will implement a loop using `PolynomialFeatures` for data transformation and `LinearRegression` for model training, then compute and store RMSE values for both sets.



````python
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import mean_squared_error
import numpy as np

# 1. Initialize empty lists to store RMSE values for training and testing sets
rmse_train_values = []
rmse_test_values = []

# 2. Loop through polynomial degrees from 1 to 10
for degree in range(1, 11):
    print(f"\n--- Processing Polynomial Degree: {degree} ---")

    # a. Initialize a PolynomialFeatures object with the current degree
    poly_features = PolynomialFeatures(degree=degree)

    # b. Transform X_train and X_test
    X_train_poly = poly_features.fit_transform(X_train)
    X_test_poly = poly_features.transform(X_test)

    # c. Initialize a LinearRegression object
    model = LinearRegression()

    # d. Train the LinearRegression model on X_train_poly and y_train
    model.fit(X_train_poly, y_train)

    # e. Make predictions on X_train_poly and X_test_poly
    y_train_pred = model.predict(X_train_poly)
    y_test_pred = model.predict(X_test_poly)

    # f. Calculate the Root Mean Squared Error (RMSE) for training set
    rmse_train = np.sqrt(mean_squared_error(y_train, y_train_pred))
    print(f"  RMSE (Train, Degree {degree}): {rmse_train:.2f}")

    # g. Calculate the Root Mean Squared Error (RMSE) for test set
    rmse_test = np.sqrt(mean_squared_error(y_test, y_test_pred))
    print(f"  RMSE (Test, Degree {degree}): {rmse_test:.2f}")

    # h. Add the calculated RMSE values to their respective lists
    rmse_train_values.append(rmse_train)
    rmse_test_values.append(rmse_test)

print("\nAll RMSE training values:", rmse_train_values)
print("All RMSE testing values:", rmse_test_values)
````

5.  **Mục đích của việc so sánh các mô hình hồi quy đa thức với các bậc khác nhau:**
    Mục đích chính là tìm kiếm sự cân bằng tối ưu giữa độ phức tạp của mô hình và khả năng khái quát hóa của nó (còn gọi là sự đánh đổi giữa độ chệch (bias) và phương sai (variance)).
    -   **Mô hình đơn giản (bậc thấp):** Có độ chệch cao (có thể không đủ linh hoạt) nhưng phương sai thấp (ít nhạy cảm với dữ liệu huấn luyện).
    -   **Mô hình phức tạp (bậc cao):** Có độ chệch thấp (có thể khớp tốt với dữ liệu huấn luyện) nhưng phương sai cao (rất nhạy cảm với dữ liệu huấn luyện và có thể hoạt động kém trên dữ liệu mới).
    Chúng ta muốn tìm bậc mà mô hình không quá đơn giản (dẫn đến underfitting) cũng không quá phức tạp (dẫn đến overfitting). Trong phân tích này, bậc 1 đã đạt được sự cân bằng này tốt nhất trên dữ liệu kiểm thử.
